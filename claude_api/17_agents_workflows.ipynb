## Workflows
A series of calls to claude meant to solve a specific problem through a predetermined series of steps

- We use these when:
    - You can picture the exact flow or steps that Claude should go through to solve a problem
    - The UX in your app constraints a user to a set of tasks


## Agents
Claude is given a goal and a set of tools. Claude is expected to figure out how to complete the goal through the provided tools

- We use these when:
    - You're not sure exactly what task or task parameters will give to claude

Image to cad workflow

![alt text](images/wokflow_image_to_cad.jpg)

For example, this app has these steps:
Here's how the workflow breaks down:

- Feed an image into Claude, asking it to describe the object
- Based on the description, ask Claude to use the CadQuery library to model the object
- Create a rendering
- Ask Claude to grade the rendering against the original image. If there are issues, fix them


## Workflow pattern: Evaluator - optimizer

![alt text](images/wokflow_pattern_evaluator_optimizer.jpg)

- Producer: Takes input and creates output (Claude using CadQuery to model the part and create a rendering)
- Grader: Evaluates the output against some criteria
- Feedback loop: If the grader doesn't accept the output, feedback goes back to the producer for improvement
- Iteration: The cycle repeats until the grader accepts the output

## Why learn workflow patterns?
Because you can have recipes for implementing your own features. The evaluator-optimizer is one workflow pattern that has worked well for ther engineers - consider using it you rown app

## Parallelization
How it works?
- Send the same image to claude multiple times simultaneously
- Each request includes specialized criteria for one material (metal criteria, polymer criteria, ceramic criteria, etc)
- Claude evaluates the part's suitability for each material independently
- Collect all the analysis results and feed them into a final aggregation step

## How this works?
Parallelization patterns flows a simple structure
- Split a single task into multiple sub-tasks
- Run the sub-tasks in parallel
- Aggregate the results together
- The parallelized sub-tasks don't need to be identical - each can have specializaed prompt, set of tools, etc

## Chaining workflows
Might seem obvios but one of the most useful, specially in complex tasks or long prompts



## What is?
Chaining workflow breaks down a large, complex task into smaller, sequential tasks. You split the work into focused steps that build on each other

![altimg](images/wokflow_chaining.jpg)

## Example
You are building a social media scrapper that creates videos and post them automatically. You could break the steps like this:

- Find related trending topics in twitter
- Select the most interesting topic (using claude)
- Research the topic (claude)
- Write a script for a short format video (claude)
- Use an AI avatar and text to speech to create a video
- Post the video in social media

![altimg](images/wokflow_chaining_deep_dive.jpg)

## Why chain istead of one big prompt?
- Split large tasks into smaller, non parallelizable subtasks, FOCUS
- Optionally do non-LLM processing between each tasks (api request or processing)
- Keep claude focused on one aspect of the overall task

## The long prompt problem
Here's where chaining becomes really valuable. You'll often encounter situations where you need Claude to write content with many specific constrains. Lets say you want Claude to write a technical article, and you specify that it should:

- Not mention that it's written by an AI
- Avoid using emojis
- Skip clichéd or overly casual language
- Write in a professional, technical tone

Even with these constrains crelay stated, claude might still produce content that violates some of your rules. You might get back an article that still uses emojis, mentions AI autorship, or sounds unprofessional

## The chanining solution
Instead of massive prompt

Step 1: Send initial prompt doesnt matter if result might not be perfect. Claude generate an artcle, but it might violate some of your constraints.
Step 2: Make a follow-up requesrt that focuses specifically on fixing the issues

## When to use chaining
Chaining workflows are cool when:
- complex tasks with multiple requirements
- claude consistenly ignores some contrainsts in long prompts
- You need to process or validate outputs between steps
- You want to keep each interaction focused and manageable

## Routing workflows
Solve a common problem in AI apps: different types of user requests need different handling approaches. Instead of using one-size-fits-all prompt, you can categorize incoming requests and route them to specializaed processing pipelines

## The problem with generic prompts
Consider a social media marketing tool that generates video scripts form user topcis. A user might enter "programming" or "surfing" as their topic, but these should produce very different types of content

## Setting up categories

In this tool we need to define the multiple types of content that we want to generate.

- Enterainment
- Educational
- Comedy
- personal vlog
...

So we provide a script generation prompt for each of them

## How routing works in practice?
The routing process happens in two steps:
1. Categorization - Sends the user's topic to Claude with a request to categorize it into one of your predefined genres
2. Specialized processing - Use the category result to select the appropiate prompt template and generate content

![altimg](images/wokflow_routing_deep_dive.jpg)

Where
- User input goes to a router component first
- The router categorizes the request using an intial Claude call
- Based on the category, the input gets forwarded to one specific rpocessing pipeline
- Each pipeline can have its own workflow, prompts, or tools optimized for that category

## When you use routing?
Routing workflows work well when:
- Your application handels diverse types of requests that need different approaches
- You can crearly define categories that cover your use cases
- The categorization step can be hnadled reliable by Claude
- The performance benefit of specialized processing outweighs the overhead of the routing step

## Agents
represent a shift from the structured workflows. While workflows are perfect when you now exactly step by step. Agents instead of defining a rigid sequence, you give claude a goal and a set of tools, then let it figure out how to combine those tools to achieve the objective

![altimg](images/agent_deep_dvie.jpg)

This flexibility makes agents attractive to building apps that handle varied, unpredictable tasks

Agents have goal and tools

## Tools should be abstract
The key insight for building effective agents is providing reasonably abstract tools rather than hyper-specialized ones. Claude Code demonstrates this principale perfectly

![altimg](images/agent_claude_code_tools.jpg)

## Best practice: Combinable tools
When designing agents, provide tools that claude can combine in creative ways, For example, a social media video agent might include

- bash - access to ffmpeg, cli tools, etc
- generate_image: generate images given a prompt
- tts : convert text into video
- post_media: post texts, images, etc to social media

## System prompts for environment inspection
You can guide claude to inspect is environment through system prompts. For complex tasks like video generation, this becames especially importnat

![altimg](images/agent_environment_inspection.jpg)

Consider a video creation agent that needs to:

- Generate video content using tools like FFmpeg
- Verify that audio dialogue is placed correctly
- Check that visual elements appear as expected

You might include system prompt instructions like:

- Use the bash tool to run whisper.cpp and generate caption files with timestamps to verify dialogue placement
- Use FFmpeg to extract screenshots from the video at regular intervals to visually inspect the output
- Compare the generated content against the original requirements

## Benefits of environment inspection
- Better progress tracking
- Error handling
- Quality assurance
- Adaptative behavior

## Practical implementation
When designing agents, always ask
"HOW WILL CLAUDE KNOW IF THIS ACTION WORKED?"
This might mean:

- Reading file contents before modifications
- Taking screenshots after UI interactions
- Checking API responses for expected data
- Validating generated content against requirements


Environment inspection transforms Claude from a blind executor of commands into an agent that can truly understand and adapt to its working environment.

## Agents vs workflows

Benefits of Workflows
- Claude can focus on one subtask at a time, generally leading to higher accuracy
- Far easier to evaluate and test, since you know each exact step
- More predictable and reliable execution
Better suited for solving specific, well-defined problems

Benefits of Agents
- Allow for more flexible user experience
Far more flexible task completion - Claude can combine tools in unexpected ways to complete a wide variety of tasks
- Can handle novel situations that weren't anticipated during development
- Can ask users for additional input when needed

Downsides of workflows
- Far less flexible
- Generally more constrained UX
- Require more upfront planning and design work

Downsides of agents
- Lower succesfull task completion rate
- More challenging to instrument, test, and evaluate
- Less predictable behavior

## When to use each approach
The always recommendations is always focus on implementing workflows where possible, and only resort agents when they are truly required. Workflows provide the reliability and predictability that most production applications need, while agents offer flexibility for scenarios where the exact requirements can't be predetermined

Consider workflows when you have well-defined processes and agents when you need to handle unpredictable, varied user requests that require creative problem-solving
