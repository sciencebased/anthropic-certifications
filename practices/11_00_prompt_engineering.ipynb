## Goal
Write a prompt that generates a daschund meal plan based in their age, weight, health and dietary restrictions

## Sample input
Age : 5 years
Weight : 10 kg
Healt : Good
Dietary restrictions: Lactose intolerance

## Ideal output
Here is a one day meal plan for a dashchund aiming to mantain weight and improve health

Daily caloric total: Approximately 1000 calories (calculated based on height, weight, and goal)

Macronutrient breakdown:
- Protein: 60 gr
- Carbohydrates: 120 gr
- Fats: 20 gr

Meal plan:
- Meal 1 (7:00 AM):
    - Chicken breast: 20gr
    - Lactose free crockets
    - Berries: 20 gr 
...
